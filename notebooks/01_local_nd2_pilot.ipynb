## Guardrails for next work

- Use fluorescence ND2 files for this pilot; do not default to BrightFocus/brightfield images.
- The four-condition block repeats alphabetically across rows/replicates.
- C/G/K/... wells: PLD3 only; no TMEM106B and no mCherry.
- D/H/L/... wells: PLD3 + TMEM106B; no mCherry.
- E/I/M/... wells: PLD3 + mCherry; mCherry reporter-control wells.
- F/J/N/... wells: PLD3 + TMEM106B + mCherry; primary experimental wells.
- Only E/I/M-phase and F/J/N-phase wells are valid for mCherry punctation-versus-diffusion analysis.
- C/G/K-phase and D/H/L-phase wells should not be treated as zero-puncta mCherry samples.
- Use fluorescence channels only.
- Do not use raw mCherry puncta as the main registration target.
- ND2 is the expected raw format; use lazy/indexed ND2 reads and save only selected pilot outputs.
- Do not batch-convert or process the full ND2 dataset without a separate approval step.
